```
def get_prime_exponents(n):
    """Get prime factorization as dictionary of prime:exponent pairs"""
    factors = {}
    d = 2
    while n > 1:
        count = 0
        while n % d == 0:
            count += 1
            n //= d
        if count > 0:
            factors[d] = count
        d += 1
        if d * d > n:
            if n > 1:
                factors[n] = 1
            break
    return factors

def analyze_feedback_loop(n, max_steps=50):
    """Analyze how prime exponents change through the sequence"""
    current = n
    steps = []
    exponent_changes = []
    
    prev_factors = get_prime_exponents(current)
    
    while current != 1 and len(steps) < max_steps:
        steps.append(current)
        
        # Store previous state
        old_factors = prev_factors.copy()
        
        # Calculate next number
        if current % 2 == 0:
            current = current // 2
        else:
            current = 3 * current + 1
            
        # Get new prime factorization
        prev_factors = get_prime_exponents(current)
        
        # Track changes in exponents
        changes = {
            'number': current,
            'type': 'even' if steps[-1] % 2 == 0 else 'odd',
            'removed_primes': {p: e for p, e in old_factors.items() if p not in prev_factors},
            'added_primes': {p: e for p, e in prev_factors.items() if p not in old_factors},
            'changed_exponents': {
                p: {'from': old_factors.get(p, 0), 'to': prev_factors.get(p, 0)}
                for p in set(old_factors) | set(prev_factors)
                if old_factors.get(p, 0) != prev_factors.get(p, 0)
            }
        }
        exponent_changes.append(changes)
    
    return exponent_changes

# Analyze a few numbers to see the feedback patterns
test_numbers = [7, 12]
for start_num in test_numbers:
    print(f"\nAnalyzing feedback loop for {start_num}:")
    changes = analyze_feedback_loop(start_num)
    
    for step, change in enumerate(changes[:5]):  # Show first 5 steps
        print(f"\nStep {step + 1}:")
        print(f"Number: {change['number']} ({change['type']} step)")
        if change['removed_primes']:
            print("Removed primes:", change['removed_primes'])
        if change['added_primes']:
            print("Added primes:", change['added_primes'])
        if change['changed_exponents']:
            print("Exponent changes:", change['changed_exponents'])
```
